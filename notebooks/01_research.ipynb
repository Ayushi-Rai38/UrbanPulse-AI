# UrbanPulse AI

## Problem Statement

Municipal corporations receive thousands of citizen complaints related to potholes, garbage, road damage, and other civic issues every day. These complaints are often processed manually, which leads to delays, incorrect prioritization, and inefficient resource allocation.

UrbanPulse AI is an AI-powered municipal infrastructure monitoring platform that helps automate the complaint management process. The system can analyze uploaded images and complaint text, identify the issue, assign priority, recommend the responsible department, generate AI-based summaries for officials, and provide dashboards with insights for better decision-making.

## Objectives

1. Detect civic issues such as potholes and garbage using uploaded images.

2. Analyze complaint text using Natural Language Processing (NLP).

3. Automatically assign complaint category, severity, priority, and responsible department.

4. Generate AI-powered summaries and recommendations for municipal officials.

5. Visualize complaints on an interactive dashboard with maps and analytics.

6. Predict future complaint trends to support proactive maintenance.

7. Build an easy-to-use platform where citizens can report issues using images or text.

# Data Type Conversion

Converting the Grievance Date column from object to datetime format for time-based analysis.

```python
df["Grievance Date"] = pd.to_datetime(df["Grievance Date"])
```

```python
df.dtypes
```

# Feature Engineering

### Objective
Extract Year, Month, and Day from the Grievance Date column. These new features will help analyze complaint patterns over time.

```python
df["Year"] = df["Grievance Date"].dt.year
df["Month"] = df["Grievance Date"].dt.month
df["Day"] = df["Grievance Date"].dt.day
```

```python
df.head()
```

# Monthly Complaint Analysis

### Objective
Analyze the number of complaints received each month to identify seasonal trends and peak complaint periods.

```python
monthly = df["Month"].value_counts().sort_index()

monthly.plot(kind="bar", figsize=(8,5))

plt.title("Complaints by Month")
plt.xlabel("Month")
plt.ylabel("Number of Complaints")

plt.show()
```

# Complaint Status Analysis

### Objective

Analyze the distribution of complaint statuses to understand how many complaints are Closed, In Progress, Registered, Reopened, or marked as Non Relevant.

```python
df["Grievance Status"].value_counts()
```

```python
import matplotlib.pyplot as plt

status = df["Grievance Status"].value_counts()

status.plot(
    kind="bar",
    figsize=(8,5)
)

plt.title("Complaint Status Distribution")
plt.xlabel("Status")
plt.ylabel("Number of Complaints")
plt.xticks(rotation=30)

plt.show()
```

### Observation

- Most complaints have been successfully closed.
- A small percentage of complaints are still in progress or recently registered.
- This indicates that the municipal corporation resolves the majority of reported issues.

# Top 10 Wards with Highest Complaints

### Objective

Identify the wards that receive the highest number of complaints.

```python
ward = df["Ward Name"].value_counts().head(10)

ward.plot(
    kind="bar",
    figsize=(10,5)
)

plt.title("Top 10 Wards by Complaints")
plt.xlabel("Ward")
plt.ylabel("Number of Complaints")

plt.xticks(rotation=45)

plt.show()
```

### Observation

- A few wards generate significantly more complaints than others.
- These wards may require additional municipal resources and monitoring.

# Category vs Complaint Status

### Objective

Analyze how complaint categories are distributed across different complaint statuses to identify which types of complaints remain unresolved or are closed successfully.

```python
pd.crosstab(df["Category"], df["Grievance Status"])
```

### Observation

- Most complaint categories have a high closure rate.
- Only a small number of complaints remain in progress or reopened.
- This indicates an overall efficient complaint resolution process.

# Top Complaint Types

### Objective

Identify the most frequently reported complaint types (Sub Categories).

```python
top_sub = df["Sub Category"].value_counts().head(15)

top_sub.plot(
    kind="barh",
    figsize=(10,6)
)

plt.title("Top 15 Complaint Types")
plt.xlabel("Number of Complaints")
plt.ylabel("Complaint Type")

plt.show()
```

### Observation

- Street Light Not Working is the most frequently reported complaint.
- Garbage collection and pothole-related complaints are also common.
- These complaint types should receive higher operational priority.
