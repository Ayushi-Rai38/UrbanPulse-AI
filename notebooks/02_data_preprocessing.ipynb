# Data Preprocessing

## Objective

Prepare the complaint dataset for Machine Learning, NLP, Dashboarding, and AI model development by performing data cleaning, feature engineering, and preprocessing.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("../data/complaints/bbmp_2024.csv")

df.head()
```

```python
df.columns = (
    df.columns
    .str.lower()
    .str.strip()
    .str.replace(" ", "_")
)

df.head()
```

## Why rename columns?

Using lowercase column names with underscores improves readability and follows industry-standard naming conventions.

```python
df["grievance_date"] = pd.to_datetime(df["grievance_date"])
```

```python
df["year"] = df["grievance_date"].dt.year
df["month"] = df["grievance_date"].dt.month
df["day"] = df["grievance_date"].dt.day
df["day_name"] = df["grievance_date"].dt.day_name()
```
